```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import os

# Load the trained YOLO model
model = YOLO('runs/detect/train/weights/best.pt')

# Specify the image path
image_path = "/content/drive/MyDrive/animal1/images/val/016ef3c82183e420_jpg.rf.fc9bb5c4fa4dcb55c95a7bf8a366e18a.jpg"

# Check if the image exists
if not os.path.exists(image_path):
    print(f"Image not found at: {image_path}")
else:
    print(f"Image found at: {image_path}")

    # Load the image using OpenCV
    image = cv2.imread(image_path)

    # Check if image is loaded successfully
    if image is None:
        print(f"Failed to load image: {image_path}")
    else:
        # Make predictions
        results = model(image_path)

        # Convert the image from BGR to RGB (for matplotlib)
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Plot the image with predictions
        fig, ax = plt.subplots(figsize=(10, 10))
        ax.imshow(image_rgb)

        # Iterate through the results and draw bounding boxes with labels
        for box in results[0].boxes:
            x1, y1, x2, y2 = box.xyxy[0].numpy()  # Box coordinates
            conf = box.conf[0].numpy()  # Confidence score
            cls = int(box.cls[0].numpy())  # Class ID

            # Get the label name from the class ID
            label = model.names[cls]

            # Draw the bounding box and label
            ax.add_patch(plt.Rectangle((x1, y1), x2 - x1, y2 - y1, linewidth=2, edgecolor='red', facecolor='none'))
            ax.text(x1, y1 - 5, f'{label} {conf:.2f}', color='red', fontsize=12, bbox=dict(facecolor='white', alpha=0.7))

        # Show the image with bounding boxes and labels
        plt.axis('off')  # Hide axes
        plt.show()
```
